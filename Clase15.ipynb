1. Biología a Redes

Las redes neuronales artificiales están inspiradas en el funcionamiento del cerebro humano. En el cerebro, las neuronas reciben información, la procesan y transmiten señales a otras neuronas. En Machine Learning, este concepto se representa mediante neuronas artificiales, que reciben datos de entrada, les asignan pesos y producen una salida. La unión de muchas neuronas artificiales forma una red neuronal, capaz de aprender patrones a partir de los datos.

2. El Perceptrón

El Perceptrón es uno de los modelos más básicos de una neurona artificial y es utilizado para realizar clasificación binaria. Recibe varias entradas, las multiplica por unos pesos, suma los resultados y agrega un sesgo o bias. Después aplica una función de activación para determinar la salida. Durante el entrenamiento, el perceptrón modifica sus pesos cuando comete errores, permitiendo aprender a clasificar los datos.

Fórmula básica:

Salida = función de activación (X1×W1 + X2×W2 + Bias)

3. MLP y Retropropagación

Un MLP (Multi-Layer Perceptron) es una red neuronal formada por varias capas de neuronas: una capa de entrada, una o más capas ocultas y una capa de salida. A diferencia del perceptrón simple, puede resolver problemas más complejos y reconocer relaciones no lineales en los datos.

La retropropagación (Backpropagation) es el proceso utilizado para entrenar estas redes. Primero se calcula la salida de la red y se compara con el resultado esperado. Después se calcula el error y este se propaga hacia atrás para ajustar los pesos de las neuronas y reducir el error en las siguientes predicciones.

4. API Secuencial

La API Secuencial es una forma sencilla de construir redes neuronales utilizando capas colocadas una después de otra, formando una secuencia. Es especialmente útil cuando la información pasa directamente de una capa a la siguiente sin tener conexiones complejas.

Por ejemplo, una red puede estar formada por una capa de entrada, una capa oculta y una capa de salida. En herramientas como Keras, el modelo secuencial permite crear este tipo de redes de manera sencilla y rápida.

5. API Funcional

La API Funcional permite construir redes neuronales más flexibles y complejas que la API Secuencial. Las capas pueden conectarse de diferentes maneras, permitiendo tener múltiples entradas, múltiples salidas, conexiones entre diferentes partes de la red y estructuras más avanzadas.

Es útil cuando el modelo necesita una arquitectura que no puede representarse simplemente como una secuencia de capas. Se utiliza frecuentemente para crear modelos más especializados de Machine Learning y Deep Learning.

6. API de Subclases

La API de Subclases permite crear modelos de redes neuronales definiendo una clase personalizada. El programador puede controlar directamente cómo funciona el modelo y cómo se conectan sus diferentes componentes.

Esta opción proporciona un mayor nivel de personalización y control que la API Secuencial y la API Funcional. Es especialmente útil para desarrollar arquitecturas de redes neuronales complejas o modelos que necesitan comportamientos específicos.
