01. Bloques Fundamentales

Son los componentes básicos que forman una red neuronal profunda. Incluyen las neuronas, capas de entrada, capas ocultas, capa de salida, pesos, sesgos y funciones de activación. Estos elementos permiten que la red procese información y aprenda patrones a partir de los datos.

02. Personalización Avanzada

Es la posibilidad de adaptar una red neuronal a un problema específico mediante la configuración de sus componentes y parámetros. Permite modificar aspectos como el número de capas, cantidad de neuronas, funciones de activación, tasa de aprendizaje y métodos de entrenamiento para obtener un mejor funcionamiento.

03. Cálculo de Gradientes

Es el proceso mediante el cual se determina cómo cambia el error de la red cuando se modifican sus pesos y parámetros. Los gradientes indican la dirección en la que deben ajustarse los pesos para reducir el error durante el entrenamiento.

04. Bucles de Entrenamiento

Son los ciclos repetitivos utilizados para entrenar una red neuronal. En cada ciclo, la red recibe datos, realiza una predicción, calcula el error, obtiene los gradientes y actualiza sus pesos. Este proceso se repite varias veces hasta que la red aprende adecuadamente los patrones de los datos.

05. Funciones y Grafos

Las funciones representan las operaciones matemáticas que realiza una red neuronal para transformar los datos de entrada en resultados. Los grafos representan visualmente la conexión y el orden de estas operaciones, mostrando cómo la información pasa de una capa a otra durante el procesamiento y entrenamiento de la red.
