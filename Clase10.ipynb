REGRESIÓN LINEAL

La regresión lineal es un algoritmo de Machine Learning supervisado que se utiliza para realizar predicciones de valores numéricos. Su objetivo es encontrar una relación entre una o varias variables de entrada y una variable que queremos predecir.

La idea principal es encontrar una línea que represente de la mejor manera posible la relación entre los datos. Para lograrlo, el modelo analiza los datos conocidos y busca reducir la diferencia entre los valores reales y los valores que predice.

La fórmula básica de la regresión lineal es:

y = mx + b

Donde:

y representa el valor que queremos predecir.
x representa el dato de entrada.
m representa la pendiente de la línea.
b representa el punto donde la línea cruza el eje.

Por ejemplo, podemos utilizar una regresión lineal para predecir el precio de una vivienda según su tamaño. Si tenemos información de diferentes casas, el modelo puede aprender la relación entre los metros cuadrados y el precio para posteriormente estimar el precio de una casa nueva.

La regresión lineal funciona mejor cuando existe una relación aproximadamente lineal entre las variables.

MÉTODO DEL GRADIENTE

El método del gradiente, conocido principalmente como descenso de gradiente (Gradient Descent), es un algoritmo utilizado para encontrar los valores de los parámetros que permiten que un modelo de Machine Learning cometa la menor cantidad de errores posible.

Durante el entrenamiento, el modelo realiza predicciones y calcula qué tan diferentes son esas predicciones de los valores reales. Esta diferencia se representa mediante una función de costo o pérdida.

El descenso de gradiente intenta reducir esa función de costo modificando poco a poco los parámetros del modelo. En cada paso calcula la dirección en la que debe moverse para disminuir el error.

El proceso puede resumirse de la siguiente manera:

Se establecen valores iniciales para los parámetros.
El modelo realiza una predicción.
Se calcula el error.
Se calcula el gradiente para conocer la dirección en la que debe modificarse el modelo.
Se actualizan los parámetros.
El proceso se repite hasta obtener un error suficientemente pequeño.

Por ejemplo, en una regresión lineal, el descenso de gradiente puede modificar la pendiente y el punto de intersección de la línea hasta encontrar una línea que se ajuste mejor a los datos.

Este método es muy importante porque se utiliza para entrenar muchos modelos de Machine Learning y redes neuronales.

REGRESIÓN POLINOMIAL

La regresión polinomial es una técnica de Machine Learning que permite realizar predicciones cuando la relación entre los datos no puede representarse correctamente mediante una línea recta.

A diferencia de la regresión lineal, la regresión polinomial puede generar una curva utilizando diferentes potencias de las variables, como x², x³ o x⁴.

Una fórmula sencilla puede ser:

y = b₀ + b₁x + b₂x²

En este caso, el modelo no solamente tiene en cuenta x, sino también x². Esto permite representar relaciones más complejas.

Por ejemplo, si queremos predecir el crecimiento de una planta dependiendo del tiempo, es posible que durante las primeras semanas crezca lentamente, después rápidamente y finalmente vuelva a crecer más despacio. Esta relación podría representarse mejor mediante una curva que mediante una línea recta.

La regresión polinomial puede ser útil cuando los datos presentan una tendencia curva. Sin embargo, si se utiliza un grado polinomial demasiado alto, el modelo puede aprender demasiado los datos de entrenamiento y producir sobreajuste.

CURVAS DE APRENDIZAJE

Las curvas de aprendizaje son gráficos utilizados para analizar el comportamiento de un modelo de Machine Learning durante su entrenamiento.

Generalmente muestran cómo cambia el error o el rendimiento del modelo a medida que aumenta la cantidad de datos utilizados para entrenarlo.

Estas curvas permiten comparar el comportamiento del modelo con los datos de entrenamiento y con datos que no ha visto anteriormente.

Son especialmente útiles para identificar dos problemas:

Sobreajuste (Overfitting): ocurre cuando el modelo aprende demasiado bien los datos de entrenamiento, incluyendo detalles o características específicas que no se repiten en datos nuevos. En este caso, el modelo puede tener un rendimiento muy bueno durante el entrenamiento, pero funcionar mal cuando recibe nuevos datos.

Subajuste (Underfitting): ocurre cuando el modelo es demasiado simple para aprender correctamente los patrones de los datos. En este caso, el modelo tiene un rendimiento bajo tanto con los datos de entrenamiento como con los datos de prueba.

Las curvas de aprendizaje ayudan a determinar si es necesario aumentar la cantidad de datos, modificar el modelo, cambiar sus parámetros o utilizar técnicas como la regularización.

REGULARIZACIÓN

La regularización es una técnica utilizada en Machine Learning para reducir el riesgo de sobreajuste.

El sobreajuste ocurre cuando un modelo se adapta demasiado a los datos utilizados durante el entrenamiento y pierde capacidad para realizar buenas predicciones con datos nuevos.

La regularización agrega una penalización a la función de costo del modelo cuando sus parámetros tienen valores demasiado grandes. De esta manera, se busca que el modelo sea más sencillo y que pueda generalizar mejor.

Los dos tipos más conocidos de regularización son:

Regularización L1 o Lasso: agrega una penalización basada en el valor absoluto de los parámetros. Puede hacer que algunos parámetros lleguen a cero, lo que permite eliminar variables que tienen poca importancia para el modelo.

Regularización L2 o Ridge: agrega una penalización basada en el cuadrado de los parámetros. Reduce el tamaño de los parámetros, pero normalmente no los lleva exactamente a cero.

Por ejemplo, si un modelo tiene demasiadas variables y comienza a memorizar los datos de entrenamiento, la regularización puede ayudar a disminuir la complejidad del modelo y mejorar su rendimiento con datos nuevos.

En resumen, la regularización busca conseguir un equilibrio entre un modelo demasiado simple y uno demasiado complejo.

CLASIFICACIÓN

La clasificación es una técnica de Machine Learning supervisado utilizada para asignar datos a diferentes categorías o clases.

A diferencia de la regresión, que busca predecir valores numéricos, la clasificación busca determinar a qué grupo pertenece un determinado dato.

Por ejemplo, un modelo de clasificación puede analizar un correo electrónico y determinar si pertenece a la categoría "Spam" o "No Spam".

Otros ejemplos de clasificación son:

Determinar si una transacción bancaria es normal o fraudulenta.
Determinar si un paciente pertenece a una categoría de riesgo determinada.
Clasificar imágenes como perro, gato o automóvil.
Determinar si un cliente puede o no recibir un determinado servicio.
Clasificar opiniones como positivas, negativas o neutrales.

Existen diferentes algoritmos utilizados para realizar clasificación, entre ellos:

Regresión logística.
Árboles de decisión.
K-Nearest Neighbors (K-NN).
Support Vector Machine (SVM).
Redes neuronales.

La clasificación puede ser binaria cuando existen solamente dos categorías, como "Sí" y "No", o multiclase cuando existen más de dos categorías, como "Bajo", "Medio" y "Alto".
