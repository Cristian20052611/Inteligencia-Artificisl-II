1. Reto de los Gradientes

El reto de los gradientes ocurre durante el entrenamiento de redes neuronales profundas cuando los valores del gradiente se vuelven demasiado pequeños o demasiado grandes. Esto puede dificultar que la red aprenda correctamente. El problema más conocido es el desvanecimiento del gradiente, donde las primeras capas reciben cambios muy pequeños y aprenden lentamente.

2. Inicialización

La inicialización es el proceso de asignar valores iniciales a los pesos y sesgos de una red neuronal antes de comenzar el entrenamiento. Una buena inicialización permite que el aprendizaje sea más estable y rápido, evitando problemas como gradientes demasiado grandes o pequeños. Algunos métodos utilizados son Xavier/Glorot y He.

3. Funciones de Activación

Las funciones de activación determinan cómo una neurona transforma la información que recibe antes de pasarla a la siguiente capa. Permiten que la red neuronal aprenda relaciones complejas y no solamente relaciones lineales. Entre las más utilizadas están ReLU, Sigmoid y Tanh.

4. Normalización

La normalización es una técnica utilizada para ajustar los valores de los datos o de las activaciones durante el entrenamiento. Su objetivo es mantener los valores dentro de rangos adecuados y hacer que el aprendizaje sea más estable y rápido. Un ejemplo muy utilizado en redes profundas es Batch Normalization.

5. Transferencia

El aprendizaje por transferencia (Transfer Learning) consiste en utilizar una red neuronal que ya fue entrenada con una gran cantidad de datos y adaptar sus conocimientos a una nueva tarea. Esto permite reducir el tiempo de entrenamiento y la cantidad de datos necesarios. Es común utilizar modelos previamente entrenados para tareas de clasificación de imágenes, procesamiento de texto y reconocimiento de objetos.

6. Preentrenamiento

El preentrenamiento consiste en entrenar previamente una red neuronal utilizando un conjunto de datos grande antes de adaptarla a una tarea específica. Durante este proceso, la red aprende características y patrones generales que posteriormente pueden aprovecharse en otra tarea. Esto facilita el entrenamiento de modelos profundos cuando se dispone de pocos datos específicos.
